```python
class mydata: #create a new class
    def __init__(self, numbers): #the initializztion function, the class object takes a ist, which is numbers 
        self.data=list(numbers)  #self.data is the list you give, numbers is just view the list you give as a parameter 
    def mean(self):              #the mean function related to the class mydata
        return sum(self.data)/len(self.data)
    def std(self):               #the standard deviation function related to the class mydata
        average=self.mean()      #mean function already created could use directly now on
        variance=sum((i-average)**2 for i in self.data)/len(self.data) #variance is the square of standard deviation 
        return variance**0.5     #hence standard deviation is the square root of variance 

x=mydata([1,2,7,5])
print("Mean:", x.mean())
print("Standard deviation:", x.std()) 
```

```python
import numpy as np #use the numpy library but use np to represent numpy to save word counts, so we only need np.

def lag(a,n): #create a function that takes a numpy vector a=np.array and an integer n 
    if n<=0:
        raise ValueError("n must be an integer greater than 0") #let the person who enter the value for n to know the boundary conditions of n
    result=np.full(len(a),np.nan) #this create a list full of nan, the number of nans is controlled by len(a)
    if n<len(a): #number of nans is less than length of the array, the array should contain non nan elements       
        result[n:]=a[:-n]
        return result 
    else:        #number of nans is equal to length of the array, all of the elements are nans 
        return result #also if n>len(a), the function will return the original result, full with nans controlled by len(a)

a=np.array([1,6,8,4])
print("Original:", a)
print("Lag by 1:", lag(a,1))
print("Lag by 2:", lag(a,2))
print("All lagged:", lag(a,len(a)))
```

```python
def lagGenerator(n):         #the lagGenerator with the parameter n, which links to the parameter n in lag function 
    return lambda a:lag(a,n) #the lambda function that declares the lag function quickly 

lag_by_2=lagGenerator(2)
a=np.array([1,6,8,4])
print(lag_by_2(a))
```

```python
def filter_and_square(num_list):            #the function to convert a number list to its squarelist but only those > 0
    return([x**2 for x in num_list if x>0]) #list comprehension to generate the square list but with condition x>0

print(filter_and_square([-2,0,3,4]))
```

```python
def frequency_map(string_list):             #create the frequuency_map function 
    counts={}                               #create a blank dictionary first 
    for words in string_list:               # iterate throught the list 
        if words in counts:                 #if the item in the list has already appear in the dictionary as a key, add 1 to the key value
            counts[words] += 1
        else:                               #if the item in the list hasn't appear yet in the dictionary as a key, add 1 to the key value
            counts[words] = 1

    return counts
word_list = ["apple", "banana", "apple", "orange"]
print(frequency_map(word_list))
```
